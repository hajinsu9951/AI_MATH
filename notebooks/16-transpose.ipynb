# 16차시 · 전치행렬 — 행과 열을 바꾸면


사진을 **전치**하면 행과 열이 뒤바뀝니다. 두 번 전치하면 원래대로 돌아옵니다.

```python
# 사진을 한 장 올려 주세요(교실 PC 사진도 좋습니다).
from google.colab import files
import numpy as np
from PIL import Image
up = files.upload()
fname = list(up.keys())[0]
img = np.array(Image.open(fname).convert('L'))   # 'L' = 회색조
print('사진 크기(행, 열):', img.shape)
```

```python
imgT = img.T
print('원본  :', img.shape)
print('전치본:', imgT.shape)   # 두 수가 뒤바뀝니다
```

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots(1, 2, figsize=(9, 4))
ax[0].imshow(img, cmap='gray'); ax[0].set_title('img'); ax[0].axis('off')
ax[1].imshow(imgT, cmap='gray'); ax[1].set_title('img.T'); ax[1].axis('off')
plt.show()
```

### 두 번 전치하면?

```python
print(np.array_equal(img, imgT.T))   # True 가 나오면 (A^T)^T = A 확인
```
