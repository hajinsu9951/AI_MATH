# 6차시 · 컴퓨터는 이 열을 무엇으로 볼까 — 자료의 종류


**명목형에 평균이 나오면 경고 신호**입니다. 직접 확인해 봅시다.

```python
import pandas as pd

df = pd.DataFrame({
    '이름': ['가온', '나린', '다솜', '라온', '마루'],
    '키':   [162, 175, 158, 180, 169],
    '혈액형': ['A', 'B', 'O', 'AB', 'A'],
    '만족도': [4, 5, 3, 5, 2],
})
df
```

```python
df.dtypes        # 컴퓨터가 각 열을 무엇으로 보았는지
```

```python
df.describe()    # 평균이 계산되는 열은 어디인가?
```

### 뜻 없는 수 — 혈액형에 번호를 매기면
A=1, B=2, O=3, AB=4 로 바꾸면 평균이 **계산은 됩니다**. 그런데 그 값은 무슨 뜻일까요?

```python
df['혈액형번호'] = df['혈액형'].map({'A': 1, 'B': 2, 'O': 3, 'AB': 4})
print('혈액형번호 평균 =', df['혈액형번호'].mean())
print('→ 이 수는 무슨 뜻인가요? 한 줄로 적어 봅시다.')
```

### 정리
- **수치형** : 더하고 평균 낼 수 있음(키·만족도)
- **명목형** : 이름표일 뿐이라 평균이 뜻을 갖지 않음(혈액형)
