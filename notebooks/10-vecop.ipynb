# 10차시 · 벡터의 합·차·실수배와 크기


손으로 계산한 값을 다섯 줄로 확인합니다.

```python
import numpy as np

a = np.array([3, 1, 2])
b = np.array([1, 1, 2])

print('a + b =', a + b)
print('a - b =', a - b)
print('2a    =', 2 * a)
```

```python
print('|a| =', np.linalg.norm(a))     # 3.7416... 즉 √14
print('√14 =', np.sqrt(14))
```

### 성분이 많아져도 똑같습니다
성분 100개짜리 벡터도 같은 식으로 다룹니다.

```python
u = np.arange(1, 101)
v = np.ones(100) * 2
print('앞 5개 합:', (u + v)[:5])
print('|u| =', round(float(np.linalg.norm(u)), 3))
```
