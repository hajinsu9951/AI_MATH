# 26차시 · 오차를 재는 두 개의 자 — MSE와 MAE


제곱으로 재는 자(MSE)와 절댓값으로 재는 자(MAE)를 **한 그림에 겹쳐** 봅니다.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 2, 3, 4])
y = np.array([0, 5, 2, 8])
a = np.linspace(0, 4, 401)

L = [((y - t * x) ** 2).mean() for t in a]      # MSE
M = [np.abs(y - t * x).mean() for t in a]       # MAE

plt.plot(a, L, label='MSE')
plt.plot(a, M, label='MAE')
plt.xlabel('a'); plt.legend(); plt.show()
```

```python
print('MSE 를 가장 작게 하는 a =', round(float(a[np.argmin(L)]), 3))
print('MAE 를 가장 작게 하는 a =', round(float(a[np.argmin(M)]), 3))
```

### 튀는 값 하나를 넣어 보면
`y` 의 마지막 값을 8 에서 **30** 으로 바꾸고 다시 실행해 보세요.
두 자 중 어느 쪽이 더 크게 흔들리나요?

```python
y2 = np.array([0, 5, 2, 30])
L2 = [((y2 - t * x) ** 2).mean() for t in a]
M2 = [np.abs(y2 - t * x).mean() for t in a]
print('MSE 최적 a =', round(float(a[np.argmin(L2)]), 3))
print('MAE 최적 a =', round(float(a[np.argmin(M2)]), 3))
```
