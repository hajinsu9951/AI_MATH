# 30차시 · 기울기와 절편을 함께 갱신하기


이제 갱신할 것이 **두 개**(a, b)입니다. 오늘 위젯과 같은 식을 반복문으로 돌립니다.

```python
x = [-2, -1, 0, 1, 2]
y = [ 1,  3, 5, 7, 10]

a, b = 0.0, 0.0
k = 0.1        # 학습률
n = len(x)
```

```python
for step in range(1, 25):
    dLda = sum(-2 * x[i] * (y[i] - (a * x[i] + b)) for i in range(n)) / n
    dLdb = sum(-2 * (y[i] - (a * x[i] + b)) for i in range(n)) / n
    a -= k * dLda
    b -= k * dLdb
    if step % 4 == 0 or step == 1:
        print(f'{step:2d}번째  a = {a:.3f},  b = {b:.3f}')
```

20행쯤에서 **a ≈ 2.2, b ≈ 5.2** 근처로 모이면 오늘 위젯과 같은 결과입니다.

```python
print('최종 a =', round(a, 3), '/ 최종 b =', round(b, 3))
L = sum((y[i] - (a * x[i] + b)) ** 2 for i in range(n)) / n
print('평균제곱오차 =', round(L, 4))
```
