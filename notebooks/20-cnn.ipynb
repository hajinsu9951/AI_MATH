# 20차시 · 합성곱과 풀링을 코드로


오늘 손으로 계산한 **9칸**을 컴퓨터가 어떻게 반복하는지 봅니다.

핵심 한 줄 : `out[i, j] = np.sum(img[i:i+3, j:j+3] * kernel)`

```python
# 사진을 한 장 올려 주세요(교실 PC 사진도 좋습니다).
from google.colab import files
import numpy as np
from PIL import Image
up = files.upload()
fname = list(up.keys())[0]
img = np.array(Image.open(fname).convert('L'))   # 'L' = 회색조
print('사진 크기(행, 열):', img.shape)
```

```python
kernel = np.array([[ 0, -1,  0],
                   [-1,  5, -1],
                   [ 0, -1,  0]])   # 또렷하게 하기(sharpen)

H, W = img.shape
out = np.zeros((H - 2, W - 2))
for i in range(H - 2):
    for j in range(W - 2):
        out[i, j] = np.sum(img[i:i+3, j:j+3] * kernel)
out = np.clip(out, 0, 255)
print('결과 크기:', out.shape)
```

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots(1, 2, figsize=(9, 4))
ax[0].imshow(img, cmap='gray'); ax[0].set_title('원본'); ax[0].axis('off')
ax[1].imshow(out, cmap='gray'); ax[1].set_title('합성곱 결과'); ax[1].axis('off')
plt.show()
```

### 맥스 풀링 — 2×2 구역마다 가장 큰 값만

```python
h, w = out.shape
ph, pw = h // 2, w // 2
pool = np.zeros((ph, pw))
for i in range(ph):
    for j in range(pw):
        pool[i, j] = np.max(out[2*i:2*i+2, 2*j:2*j+2])

print('풀링 전:', out.shape, '→ 풀링 후:', pool.shape)
plt.imshow(pool, cmap='gray'); plt.axis('off'); plt.show()
```
