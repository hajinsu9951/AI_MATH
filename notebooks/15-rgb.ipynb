# 15차시 · 색을 나누고 회색으로 — RGB 세 채널


사진 한 장을 **R·G·B 세 장**으로 나누고, 회색조로 바꾸는 두 방식을 견주어 봅니다.

위 메뉴에서 **런타임 → 모두 실행**을 누르면 순서대로 돌아갑니다.

```python
from google.colab import files
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt

up = files.upload()
fname = list(up.keys())[0]
img = np.array(Image.open(fname).convert('RGB'))
print('크기(행, 열, 채널):', img.shape)
```

### 1. 세 채널로 나누기
각 채널은 0~255 사이 정수로 이루어진 **행렬 한 장**입니다.

```python
r, g, b = img[:, :, 0], img[:, :, 1], img[:, :, 2]

fig, ax = plt.subplots(1, 3, figsize=(12, 4))
for a, m, name in zip(ax, [r, g, b], ['R', 'G', 'B']):
    a.imshow(m, cmap='gray'); a.set_title(name); a.axis('off')
plt.show()
print('R 채널 왼쪽 위 3x3\n', r[:3, :3])
```

### 2. 회색조 두 가지 — 단순 평균 vs 사람 눈 가중치

- 단순 평균 : `(R+G+B)/3`
- 사람 눈 가중치 : `0.299R + 0.587G + 0.114B`  ← 초록에 가장 큰 가중치

```python
mean = (r.astype(float) + g + b) / 3
luma = 0.299 * r + 0.587 * g + 0.114 * b

fig, ax = plt.subplots(1, 2, figsize=(9, 4))
ax[0].imshow(mean, cmap='gray'); ax[0].set_title('mean'); ax[0].axis('off')
ax[1].imshow(luma, cmap='gray'); ax[1].set_title('luma'); ax[1].axis('off')
plt.show()

print('두 방식의 평균 차이:', round(float(np.abs(mean - luma).mean()), 2))
```
