# 17차시 · 행렬곱으로 사진 옮기기·뒤집기


**곱하는 자리**에 따라 결과가 달라집니다. `img @ D` 와 `D @ img` 를 견주어 보세요.

```python
# 사진을 한 장 올려 주세요(교실 PC 사진도 좋습니다).
from google.colab import files
import numpy as np
from PIL import Image
up = files.upload()
fname = list(up.keys())[0]
img = np.array(Image.open(fname).convert('L'))   # 'L' = 회색조
print('사진 크기(행, 열):', img.shape)
```

```python
n, m = img.shape
D = np.eye(m)[::-1]        # 열 순서를 뒤집는 반전 행렬
E = np.eye(n)[::-1]        # 행 순서를 뒤집는 반전 행렬
print('D 크기:', D.shape, '| E 크기:', E.shape)
```

```python
import matplotlib.pyplot as plt
left  = img @ D            # 좌우 반전
updown = E @ img           # 상하 반전

fig, ax = plt.subplots(1, 3, figsize=(13, 4))
for a, m2, t in zip(ax, [img, left, updown], ['원본', 'img @ D (좌우)', 'E @ img (상하)']):
    a.imshow(m2, cmap='gray'); a.set_title(t); a.axis('off')
plt.show()
```

### 평행이동 행렬
`np.eye(n, k=1)` 은 대각선이 한 칸 밀린 행렬입니다.

```python
S = np.eye(m, k=1)
plt.imshow(img @ S, cmap='gray'); plt.axis('off'); plt.show()
print('AB 와 BA 가 같은가?', np.array_equal(img @ D, D.T @ img.T))
```
