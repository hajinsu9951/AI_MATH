# 19차시 · 완전연결층 — 순전파 한 줄


입력 벡터에 **가중치 행렬**을 곱하면 다음 층의 값이 나옵니다. 이것이 순전파입니다.

```python
import numpy as np

X = np.array([[1, 3, 4]])
W = np.array([[1, 2, 3],
              [1, 0, 1],
              [0, 1, 1]])
print('X 꼴:', X.shape, '| W 꼴:', W.shape)
```

```python
Y = X @ W
print(Y)             # [[4 6 10]] 이 나오면 손계산과 같습니다
print('가장 큰 칸:', int(np.argmax(Y)))
```

### 은닉층을 하나 더 쌓으면
층을 더 쌓아도 **행렬곱을 한 번 더** 할 뿐입니다.

```python
W1 = W
W2 = np.array([[1, 0],
               [0, 1],
               [1, 1]])
H = X @ W1
O = H @ W2
print('은닉층 H =', H)
print('출력   O =', O)
print('판단   =', int(np.argmax(O)))
```
