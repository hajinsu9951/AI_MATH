# 31차시 · 집단별 합격률과 기준선 바꾸기


학습지의 자료를 올려 **집단별 합격률**을 구하고, 기준선을 바꿔 가며 격차를 봅니다.

```python
import pandas as pd
from google.colab import files

# CSV 를 올려 주세요. 열 이름 예시: gender, score
up = files.upload()
df = pd.read_csv(list(up.keys())[0])
df.head()
```

```python
cut = 70   # 합격 기준선
df['합격'] = df['score'] >= cut
df.groupby('gender')['합격'].mean()
```

### 기준선을 50 에서 90 까지 바꿔 보면

```python
import matplotlib.pyplot as plt

rows = []
for c in range(50, 91, 5):
    p = df.assign(합격=df['score'] >= c).groupby('gender')['합격'].mean()
    rows.append({'기준선': c, **{k: round(v, 3) for k, v in p.items()},
                 '격차': round(float(p.max() - p.min()), 3)})
tbl = pd.DataFrame(rows)
tbl
```

```python
plt.plot(tbl['기준선'], tbl['격차'], marker='o')
plt.xlabel('기준선'); plt.ylabel('집단 간 합격률 격차'); plt.show()
```

### 생각해 보기
격차가 가장 작은 기준선이 늘 가장 좋은 기준선일까요? 정확도와 함께 견주어 봅시다.
